1. Now let's define the domain:

```python
length = 2              # Define the domain dimension limits
nx = 41                 # Defines the number of points of the grid domain
dx = length / (nx - 1)  # Computes the distance between two consecutive points
nt = 25                 # Defines the number of timsteps to compute
dt = 0.025              # Defines how much each timestep covers/advances
c = 1                   # Speed variable of the wave differential equation. We assume c = 1
```

2. Let's define the initial conditions
    * The initial velocity is defined as:

$$
u(x,0) =
\begin{cases}
1, & x \in [0,0.5) \cup (1,2] \\
2, & x \in [0.5,1]
\end{cases}
$$


```python
u = np.ones(nx)
u[int(.5/dx):int(1/dx)+1] = 2
print(u)
```

3. Let's plot the array generated:

```python
plt.plot(np.linspace(0, 2, nx), u);
```

4. Let's see how to iterate through the grid and save the values:

```python
u_n = np.ones(nx) # initialize an array to save the values of every u_i of a time step

for n in range(1, nt): # iterate through the values of n (time) form 0 to nt points/times
    u_n = u.copy() # copy the values of the initial conditions
    for i in range(1, nx): # iterate through the values of i (space) from 0 to nx points/positions
         u[i] = u_n[i] - c * (dt / dx) * (u_n[i] - u_n[i-1])
    #for i in range(nx):
    #    u[i] = u_n[i] - c * (dt / dx) * (u_n[i] - u_n[i-1])
```

5. Let's plot the results and see what happens:

```python
plt.plot(np.linspace(0, 2, nx), u);
```
