# _Step 2: Nonlinear Convection_

Now let's implement the same methods used previously for solving linear convection, but this time to solve the same equation with a variable convection velocity, (u):

$$
\frac{\partial u}{\partial t} + u\frac{\partial u}{\partial x} = 0
$$

Then so, the discretized form is as follows:

$$
\frac{u_i^{n+1} - u_i^n}{\Delta t} + u_i^n \frac{u_i^n - u_{i-1}^n}{\Delta x} = 0 \quad \rightarrow \quad u_i^{n+1} = u_i^n - u_i^n \frac{\Delta t}{\Delta x} (u_i^n - u_{i-1}^n)
$$

### 1. Import and define the grid/domain dimensions and parameters:

$$
\text{Initial conditions} \quad\rightarrow\quad
u(x,0) =
\begin{cases}
1, & x \in [0,0.5) \cup (1,2] \\
2, & x \in [0.5,1]
\end{cases}
$$

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

nx = 500
dx = 2 / (nx - 1)
nt = 500 # Try values 20, 100, 500, 1000, 2000, 2600, ...
dt= .00025

u = np.ones(nx)
u[int(.5/dx) : int(1/dx+1)] = 2
plt.plot(np.linspace(0, 2, nx),u)

u_n = np.ones(nx) # temporal array

for n in range(nt):
    u_n = u.copy()
    for i in range(1, nx):
                 u[i] = u_n[i] - u_n[i] * (dt / dx) * (u_n[i] - u_n[i-1])

plt.plot(np.linspace(0, 2, nx),u)
```

### What do you observe about the evolution of the hat function?
The pulse moves to the right, but it doesn’t keep its shape, because each point travels at a speed equal to its own value of u: the part where u=2 moves twice as fast as the part where u=1. Where it goes from 1 to 2, the range of x values over which the transition happens expands (expansion wave): the fast part runs ahead and the step turns into a ramp that keeps getting wider. Where it changes from 2 to 1, the opposite happens: the fast part catches up with the slow part and the transition is compressed, so it stays as an almost vertical jump (a shock). If it’s left to evolve for longer, the ramp eventually catches up with the shock, the plateau disappears and the maximum starts to drop.

### What happens when you change the numerical parameters?
When the mesh is refined (more points in x), the shock becomes narrower and the ramp looks more like a straight line. The scheme adds numerical diffusion that smooths the edges, and that diffusion gets smaller as Δx gets smaller.
